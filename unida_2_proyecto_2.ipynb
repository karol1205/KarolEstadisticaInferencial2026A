````python
# 12. Realice una tabla ANOVA e interprete el resultado.

# test de ANOVA (Analysis of Variance)
# H0: beta_1 = 0   (No hay correlación)
# H1: beta_1 ≠ 0   (Sí hay correlación)

import statsmodels.api as sm
from statsmodels.formula.api import ols
# Y ~ X
modelo_lineal = ols(
    '''precio ~ habitaciones + banios + area''',
    data = df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal)
tabla_anova
````

````python
anova = sm.stats.anova_lm(modelo_lineal)

ss_total = anova["sum_sq"].sum()
ss_residual = anova.loc["Residual", "sum_sq"]

anova["eta_sq"] = anova["sum_sq"] / ss_total
anova["eta_sq_parcial"] = anova["sum_sq"] / (anova["sum_sq"] + ss_residual)

anova
````

````python
import statsmodels.api as sm
x_constante = sm.add_constant(x)

x_constante
````

````python
import statsmodels.api as sm
modelo = sm.OLS(y, x_constante).fit()
y_calculada = modelo.predict(x_constante)

modelo.params
````

Las pruebas de Omnibus y Jarque-Bera permiten comprobar si los residuos siguen una distribución normal. En este caso, ambas tienen valores p menores a 0.05, por lo que se rechaza la hipótesis nula y se concluye que los residuos no siguen una distribución normal. Además, el sesgo (Skew) de 0.928 y la curtosis (Kurtosis) de 4.611 muestran que los datos se alejan de una distribución normal.
En cuanto al modelo, el valor de R-squared es de 0.192, lo que significa que el área, los baños y las habitaciones explican el 19.2% de las variaciones en el precio de las propiedades. Por su parte, el estadístico F de 56.12 indica que el modelo es significativo en general.
Al analizar las variables, el área tiene un coeficiente de 607.35 y los baños de 12,910; ambas son estadísticamente significativas, ya que sus valores p son menores a 0.05. Esto indica que un aumento en cualquiera de estas variables se relaciona con un incremento en el precio, manteniendo las demás constantes. En cambio, las habitaciones no son significativas, pues su valor p es de 0.974.
Finalmente, Durbin-Watson tiene un valor de 1.865, cercano a 2, lo que sugiere que no hay una autocorrelación importante en los errores. Sin embargo, el número de condición de 1.31e+03 es relativamente alto, por lo que sería conveniente revisar si existe multicolinealidad entre las variables.

````python
residuales = modelo.resid
````

````python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuales)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuales, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuales, kde=True)
plt.title("Histograma de residuos")
plt.show()
````

````python
# 10. Realice la prueba de Breusch-Pagan para los residuales
# y comente el resultado.

# Test de Breusch-Pagan
# H0: No hay heterocedasticidad (dispersión es igual en toda la recta)
# H1: Hay Heterocedasticidad (dispersión distinta a lo largo de la recta)
# alpha = 0.05

from statsmodels.stats.api import het_breuschpagan
_, valor_p_bp, _, _ = het_breuschpagan(residuales, x_constante)
print(f'valor_p de Breusch-Pagan: {valor_p_bp: 0.4f}\n')
````
